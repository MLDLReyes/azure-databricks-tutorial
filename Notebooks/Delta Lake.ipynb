
## Managed VS External Delta Tables


**Database**

```python
%sql

CREATE DATABASE salesDB;
```


**Managed Table**

```python
%sql

CREATE TABLE salesDB.mantable(
id INT,
name STRING,
marks INT
)
USING DELTA
```


## Delta Table Functionalities


**INSERT**

```python
%sql

INSERT INTO salesDB.externaltable VALUES
(5,'aa',30),
(6,'bb',33),
(7,'cc',35),
(8,'DD',40)
```

```python
%sql

SELECT * FROM salesDB.externaltable
```


**DELETE**

```python
%sql

DELETE FROM salesDB.externaltable
WHERE id = 8
```

```python
%sql
SELECT * FROM salesDB.externaltable
```


**DATA VERSIONING**

```python
%sql

DESCRIBE HISTORY salesDB.externaltable
```


**TIME TRAVEL**

```python
%sql

RESTORE TABLE salesDB.externaltable TO VERSION AS OF 2;
```

```python
%sql

SELECT * FROM salesDB.externaltable
```


**VACUUM**



```python
%sql

--BE CAREFUL WHEN RUNNING THE VACUUM COMMAND, IT WILL DELETE ALL THE OLD PARQUET FILES WHICH WILL MAKE YOU UNABLE TO PERFORM TIME TRAVEL/VERSIONING

--VACUUM ONLY DELETES PARQUET FILES THAT ARE AT LEAST 7 DAYS OLD FOR THE DEFAULT TIME FRAME

--VACUUM salesDB.externattable
```


**VACUUM RETAIN 0 HOURS**

```python
%sql

--TO DELETE REGARDLESS OF TIME FRAME

--VACUUM salesDB.externattable RETAIN 0 HOURS
```
