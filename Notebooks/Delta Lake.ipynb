
## Managed VS External Delta Tables


**Database**

```python
%sql

CREATE DATABASE salesDB;
```


**Managed Table**

```python
%sql

CREATE TABLE salesDB.mantable(
id INT,
name STRING,
marks INT
)
USING DELTA
```

```python
%sql

INSERT INTO salesDB.mantable VALUES
(1,'aa',30),
(2,'bb',33),
(3,'cc',35),
(4,'DD',40)
```

```python
%sql

SELECT * FROM salesDB.mantable
```

```python
%sql

DROP TABLE salesdb.mantable;
```


**Setting Up External Location**

**External Table**

```python
%sql

CREATE TABLE salesDB.externaltable (
  id INT,
  name STRING,
  marks INT
)
USING DELTA
LOCATION 'abfss://destination@datalakemldlreyes.dfs.core.windows.net/salesDB/externaltable';
```

```python
%sql

INSERT INTO salesDB.externaltable VALUES
(1,'aa',30),
(2,'bb',33),
(3,'cc',35),
(4,'DD',40)
```

```python
%sql

SELECT * FROM salesdb.externaltable;
```
